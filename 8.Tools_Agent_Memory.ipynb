```python
# 🧠 ConversationBufferMemory Example

from langchain_groq import ChatGroq
from langchain.prompts import PromptTemplate
from langchain.chains import LLMChain
from langchain.tools import Tool
from langchain.memory import ConversationBufferMemory
from langchain.agents import initialize_agent, AgentType

# 🔑 Groq API Key

groq_api_key = "API KEY"

# 🔸 Initialize LLM

llm = ChatGroq(
    groq_api_key=groq_api_key,
    model_name="llama-3.1-8b-instant",
    temperature=0
)

# ✅ Simple QA Tool

qa_prompt = PromptTemplate(
    input_variables=["question"],
    template="Answer clearly: {question}"
)

qa_chain = LLMChain(
    llm=llm,
    prompt=qa_prompt
)

qa_tool = Tool(
    name="Simple_QA",
    func=lambda q: qa_chain.run(question=q),
    description="Answer factual questions clearly"
)

# 🧠 Memory

memory = ConversationBufferMemory(
    memory_key="chat_history",
    return_messages=True
)

# 🤖 Agent with Memory

agent = initialize_agent(
    tools=[qa_tool],
    llm=llm,
    agent=AgentType.CONVERSATIONAL_REACT_DESCRIPTION,
    verbose=True,
    memory=memory,
    handle_parsing_errors=True
)

# 🚀 Multiple Interactions

print("1️⃣ First Question")

res1 = agent.run("What is LangChain?")

print("\nAnswer:")
print(res1)

print("\n2️⃣ Follow-up Question")

res2 = agent.run("Who created it?")

print("\nAnswer:")
print(res2)

print("\n3️⃣ Ask again about previous topic")

res3 = agent.run("Explain it simply again.")

print("\nAnswer:")
print(res3)

# 🧾 Chat Memory History

print("\n🧠 Conversation History:\n")

for msg in memory.chat_memory.messages:
    print(f"{msg.type.upper()}: {msg.content}")
```

```python
# 🧠 VectorStoreRetrieverMemory Example

from langchain_groq import ChatGroq
from langchain.prompts import PromptTemplate
from langchain.chains import LLMChain
from langchain.tools import Tool
from langchain.memory import VectorStoreRetrieverMemory
from langchain.agents import initialize_agent, AgentType

from langchain_community.vectorstores import FAISS
from langchain_community.embeddings import FakeEmbeddings

# 🔑 Groq API Key

groq_api_key = "API KEY"

# 🔸 Initialize Groq LLM

llm = ChatGroq(
    groq_api_key=groq_api_key,
    model_name="llama-3.1-8b-instant",
    temperature=0
)

# ✅ Simple QA Tool

qa_prompt = PromptTemplate(
    input_variables=["question"],
    template="Answer clearly: {question}"
)

qa_chain = LLMChain(
    llm=llm,
    prompt=qa_prompt
)

qa_tool = Tool(
    name="Simple_QA",
    func=lambda q: qa_chain.run(question=q),
    description="Answer factual questions clearly"
)

# 🧠 Fake Embeddings (No OpenAI API Needed)

embedding = FakeEmbeddings(size=1352)

# 🗂️ Create FAISS Vector Store

vectorstore = FAISS.from_texts(
    ["initial memory"],
    embedding
)

retriever = vectorstore.as_retriever()

# 🧠 Vector Memory

memory = VectorStoreRetrieverMemory(
    retriever=retriever,
    memory_key="chat_history"
)

# 🤖 Agent with Memory

agent = initialize_agent(
    tools=[qa_tool],
    llm=llm,
    agent=AgentType.CONVERSATIONAL_REACT_DESCRIPTION,
    verbose=True,
    memory=memory
)

# 🚀 Multiple Interactions

print("1️⃣ First Question")

res1 = agent.run("What is LangChain?")

print("\nAnswer:")
print(res1)

print("\n2️⃣ Follow-up Question")

res2 = agent.run("Who created it?")

print("\nAnswer:")
print(res2)

print("\n3️⃣ Ask again about previous topic")

res3 = agent.run("Explain it simply again.")

print("\nAnswer:")
print(res3)
```

```python
# 🔍 Check what is stored in FAISS retriever

docs = retriever.vectorstore.similarity_search("memory")

print("\n📚 Stored Memory Documents:\n")

for i, doc in enumerate(docs):
    print(f"🔹 Document {i+1}:")
    print(doc.page_content)
    print("-" * 40)
```

```python
# 📚 See all texts stored inside FAISS Vector Store

print("All indexed texts:\n")

for i, doc in enumerate(retriever.vectorstore.docstore._dict.values()):
    print(f"📄 Document {i+1}:")
    print(doc.page_content)
    print("-" * 40)
```

```python
# 🔎 Retrieve relevant memory documents

query = "LangChain founder"

results = memory.retriever.get_relevant_documents(query)

print(f"\n🧠 Retrieval for: '{query}'\n")

for i, doc in enumerate(results):
    print(f"🔹 Result {i+1}:")
    print(doc.page_content)
    print("-" * 40)
```
