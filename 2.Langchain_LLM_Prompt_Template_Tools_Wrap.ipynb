```python
from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate
from langchain_core.tools import Tool

# 🔑 Groq API Key
groq_api_key = "Your API key"

# 🔸 Step 1: Initialize Groq LLM
llm = ChatGroq(
    groq_api_key=groq_api_key,
    model_name="llama-3.1-8b-instant",
    temperature=0
)

# 🔸 Step 2: Create Prompt Template
prompt = PromptTemplate.from_template(
    "Answer clearly: {question}"
)

# 🔸 Step 3: Create Chain
chain = prompt | llm

# 🔸 Step 4: Create Tool Function
def qa_function(question):
    response = chain.invoke({
        "question": question
    })
    return response.content

# 🔸 Step 5: Wrap as Tool
qa_tool = Tool(
    name="Simple QA",
    func=qa_function,
    description="A basic LLM tool that answers clear questions"
)

# 🔸 Step 6: Use Tool
query = "What is LangGraph in LangChain?"

answer = qa_tool.invoke(query)

# 🖨️ Output
print("\nUser Question:", query)
print("\nTool Answer:\n")
print(answer)
```
